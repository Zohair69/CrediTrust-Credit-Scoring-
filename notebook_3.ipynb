### Description des colonnes

Informations personnelles

- Sexe (_Gender_) : indique si le demandeur est un homme ou une femme.
- État civil (_Married_) : indique l'état civil du demandeur (Oui ou Non).
- Personnes à charge (_Dependents_) : nombre de personnes dont le demandeur assure la prise en charge financière (par exemple, 0, 1, 2, 3 ou plus).
- Formation (_Education_): niveau d'études du demandeur (diplômé de l'enseignement supérieur ou étudiant de premier cycle).
- Indépendant (_Self_Employed_) : indique si le demandeur exerce une activité indépendante (Oui ou Non). 

Informations financières 

- Revenu du demandeur (_ApplicantIncome_): montant des revenus du demandeur principal.
- Revenu du codemandeur (_CoapplicantIncome_): montant des revenus du cosignataire ou du partenaire.
- Montant du prêt (_LoanAmpount_): montant total demandé pour le prêt (généralement en milliers).
- Durée du prêt (Loan_Amount_Term  ): durée accordée pour rembourser le prêt (généralement en mois, par exemple 360 pour 30 ans).
- Historique de crédit (_Credit_History_): un historique des antécédents de remboursement des dettes (1 signifie un bon crédit, 0 signifie un mauvais crédit). 

Localisation et résultat 

- Zone de la propriété  (_Property_Area_) : le type de zone où se trouve la propriété (urbaine, périurbaine ou rurale).
- Statut du prêt (_Loan_Status_) : la décision finale concernant le prêt (Y pour « approuvé », N pour « refusé »). Il s’agit de la variable cible que vous essayez de prédire.

## Encodage des variables catégorielles

```python
## Encodage des variables catégorielles
colonnes_a_encoder = ["Gender", "Married", "Dependents", "Education", "Self_Employed", "Property_Area"]

X_train = pd.get_dummies(X_train, columns=colonnes_a_encoder, drop_first=True)
X_test = pd.get_dummies(X_test, columns=colonnes_a_encoder, drop_first=True)
```

**Commentaire :** aucun des modèles ne peut travailler directement avec du texte (`"Male"`, `"Yes"`, `"Urban"`...). 
On transforme donc chaque variable catégorielle en colonnes numériques 0/1 via `pd.get_dummies`. On utilise `drop_first=True` pour éviter la redondance : avec 2 catégories (ex. `Married`), une seule colonne (`Married_Yes`) suffit à représenter l'information. Si elle vaut 0, on sait que c'est "No" sans avoir besoin d'une deuxième colonne.

## Vérification de la cohérence des colonnes après encodage

```python
print(X_train.shape)
print(X_test.shape)

colonnes_manquantes_train = set(X_test.columns) - set(X_train.columns)
colonnes_manquantes_test = set(X_train.columns) - set(X_test.columns)

print("Colonnes dans test mais pas dans train :", colonnes_manquantes_train)
print("Colonnes dans train mais pas dans test :", colonnes_manquantes_test)
```

**Commentaire :** `X_train` et `X_test` ont été encodés séparément, ce qui peut créer un décalage si une catégorie rare n'apparaît que dans l'un des deux jeux de données. Ici, `Married_No` s'est retrouvée uniquement dans `X_train` (18 colonnes) et absente de `X_test` (17 colonnes). un modèle entraîné sur 18 colonnes ne peut pas prédire sur un tableau à 17.

## Correction de l'incoherance des colonnes après encodage 

```python
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)

print(X_train.shape)
print(X_test.shape)
```

**Commentaire :** `align()` force `X_test` à avoir exactement les mêmes colonnes que `X_train`, dans le même ordre, en complétant la colonne manquante (`Married_No`) avec des 0. Les deux jeux ont maintenant 18 colonnes chacun, prêts pour l'entraînement.

## Choix des modèles de classification

Pour ce projet de scoring de crédit, deux contraintes nous orientent directement le choix des modèles :

1. **La réduction des Faux Négatifs est prioritaire** : un mauvais payeur classé à tort comme "bon payeur" (Faux Négatif) coûte plus cher à la banque qu'un bon payeur refusé par erreur (Faux Positif). 

2. **L'interprétabilité est exigée**  : dans un contexte bancaire réglementé, un refus de crédit doit pouvoir être expliqué au client. Cela exclut d'office les modèles dits "boîte noire", c'est-à-dire les modèles qui produisent une prédiction sans qu'on puisse reconstruire le raisonnement qui y mène (l'inverse d'un modèle interprétable, où chaque variable a un poids ou une règle qu'on peut lire et expliquer).

C'est pourquoi nous avons retenu 3 modèles, tous interprétables, mais avec des logiques complémentaires :

### 1. Régression Logistique
Modèle de référence ("baseline") : rapide à entraîner, et chaque variable obtient un coefficient clair : un chiffre positif ou négatif qui indique directement dans quel sens et avec quelle intensité elle influence la probabilité d'accord ou de refus. 

### 2. Arbre de Décision
Le modèle le plus transparent des trois : sa structure entière peut être visualisée comme une suite de questions successives, ce qui permet d'expliquer une décision individuelle à un client de façon très concrète. En contrepartie, un arbre non limité en profondeur risque de sur-apprendre les données d'entraînement.

### 3. Random Forest
Une moyenne de nombreux arbres de décision, généralement plus performante et plus robuste qu'un arbre seul, tout en restant interprétable grâce à `feature_importances_`, qui indique quelles variables pèsent le plus dans les décisions du modèle, en moyenne sur l'ensemble des arbres. C'est un bon compromis entre performance et transparence.

**Modèles écartés :** KNN a été exclu car il n'offre aucune interprétabilité (pas de coefficients ni d'importance de variables) et gère mal le mélange de variables catégorielles et numériques présent dans ce dataset. Le SVM a également été écarté, pour les mêmes raisons de faible interprétabilité par défaut.

## Random Forest (Zohair)

**Pourquoi ce modèle n'a pas besoin de standardisation :** la Random Forest ne calcule pas de distances entre points (contrairement à KNN ou la régression logistique), donc l'échelle des variables n'a pas d'impact. on utilise `X_train`/`X_test` bruts, pas de version `_scaled`. On garde `random_state=42` pour que le résultat soit reproductible d'une exécution à l'autre.

```python
from sklearn.ensemble import RandomForestClassifier

model_rf = RandomForestClassifier(random_state=42)
model_rf.fit(X_train, y_train)

y_pred_rf = model_rf.predict(X_test)
y_pred_rf
```

**Pourquoi regarder le rappel par classe, pas juste l'accuracy :** Avec mon équipe, on a décidé de prioriser la réduction des Faux Négatifs. Avec une accuracy de 0.84, le modèle a l'air globalement bon. mais le détail par classe révèle que le rappel sur N (refusé/à risque) n'est que de 0.63 : sur 38 vrais mauvais payeurs, 14 ont été classés à tort comme "Y" (bon payeur). C'est cette erreur précise, la plus coûteuse pour la banque, qu'il faut surveiller en priorité et l'accuracy seule l'aurait masquée.

## Random Forest  avec class_weight balanced (comparaison)

```python
model_rf_balanced = RandomForestClassifier(random_state=42, class_weight='balanced')
model_rf_balanced.fit(X_train, y_train)
y_pred_rf_balanced = model_rf_balanced.predict(X_test)

print(classification_report(y_test, y_pred_rf_balanced))
```

**Pourquoi tester cette variante :** face au déséquilibre des classes (68.7% Y / 31.3% N) et notre  priorité  sur les Faux Négatifs, `class_weight='balanced'` pénalise davantage les erreurs sur la classe minoritaire (N) pendant l'entraînement. Résultat : le rappel sur N monte de 0.63 à 0.68 (moins de mauvais payeurs ratés), au prix d'une légère baisse de précision sur N (0.80 → 0.76).
Le modèle devient un peu plus "méfiant", donc accuse à tort un peu plus de bons payeurs. 
L'accuracy globale reste identique (0.84) dans les deux cas, ce qui confirme qu'elle ne suffit pas à elle seule pour départager ces deux versions.

### Choix du modèle final : Random Forest (balanced)

Le brief priorise la réduction des faux négatifs : un mauvais payeur non détecté coûte plus cher à la banque qu'un bon payeur refusé à tort. Le **rappel sur la classe N** est donc le critère de choix principal, pas l'accuracy globale.

Avec un rappel classe N de **0.684**, contre 0.579 (Régression Logistique), 0.605 (Arbre de Décision) et 0.632 (Random Forest standard), le Random Forest balanced est le modèle qui détecte le mieux les mauvais payeurs. même si l' accuracy est légèrement inférieure (0.837 vs 0.862 pour la Régression Logistique). C'est un compromis assumé par notre équipe, cohérent avec l'objectif métier.
